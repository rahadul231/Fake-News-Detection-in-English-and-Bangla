## Experiment Overview

This study investigates automated fake news detection in English and Bangla using classical machine learning and transformer-based natural language processing approaches.

The final internal experiment uses 40,064 news records after duplicate removal, comprising:

- English: 20,024 records
- Bangla: 20,040 records

For each language, 14,000 records are used for training and 4,000 for internal testing. The processed validation datasets contain 2,024 English and 2,040 Bangla records.

Four classical machine learning classifiers are evaluated using language-specific TF-IDF representations:

1. Logistic Regression
2. Multinomial Naive Bayes
3. Linear Support Vector Machine (SVM)
4. Random Forest

Transformer experiments include BanglaBERT for Bangla and multilingual BERT (mBERT) and XLM-RoBERTa for both languages.

Classical model selection is performed using validation weighted F1-score, while held-out test performance is reported separately for all evaluated models.

Additional external evaluation of the validation-selected Logistic Regression models is conducted using 68,002 English and 18,150 Bangla news records.

The study examines classification effectiveness, differences between languages, statistical comparisons, error patterns, and cross-dataset generalization.

## Introduction

The rapid growth of online news platforms and social media has increased the accessibility and circulation of news content. Although English-language fake news detection has benefited from extensive research and relatively abundant NLP resources, Bangla presents additional challenges related to available annotated data, morphological variation, vocabulary, and linguistic representation.

This study investigates fake news detection in English and Bangla using separate language-specific datasets and classification pipelines. Four classical machine learning algorithms are evaluated using TF-IDF representations, while BanglaBERT, multilingual BERT, and XLM-RoBERTa are investigated as transformer-based alternatives.

The study compares internal validation and testing performance across models and languages. Additional external evaluation of selected classical models examines the extent to which performance generalizes beyond the original experimental datasets.

The objective is to evaluate the strengths and limitations of established classification approaches while identifying methodological challenges associated with dataset quality, preprocessing consistency, model selection, and generalization.

```python
# Step 9: Basic Data Type Standardization

import pandas as pd
import numpy as np

all_datasets = [
    english_train,
    english_validation,
    english_test,
    bangla_train,
    bangla_validation,
    bangla_test
]

dataset_names = [
    "English Train", "English Validation", "English Test",
    "Bangla Train", "Bangla Validation", "Bangla Test"
]

for name, df in zip(dataset_names, all_datasets):
    labels = pd.to_numeric(df["label"], errors="coerce")
    valid_labels = labels.isin([0, 1])
    valid_text = (
        df["text"]
        .astype("string")
        .str.strip()
        .ne("")
        .fillna(False)
    )

    invalid_rows = ~(valid_labels & valid_text)

    if invalid_rows.any():
        invalid_count = int(invalid_rows.sum())
        print(f"{name}: {invalid_count} invalid rows found.")

        # Save invalid records for manual verification
        audit_file = (name.lower().replace(" ", "_") + "_invalid_rows.csv")

        df.loc[invalid_rows, ["text", "label"]].to_csv(
            audit_file, index=False
        )

        print(f"Invalid records saved to: {audit_file}")

        # Do not automatically remove training or test records
        if "Validation" not in name:
            raise ValueError(
                f"Invalid records found in {name}. "
                "Please repair the original CSV."
            )

        # Exclude invalid validation rows from this run
        df.drop(
            index=df.index[invalid_rows],
            inplace=True
        )

    df["text"] = df["text"].astype(str)

    df["label"] = (
        pd.to_numeric(df["label"])
        .astype(int)
    )

    df.reset_index(drop=True, inplace=True)

# Check and remove duplicate validation articles

for name, df in [
    ("English Validation", english_validation),
    ("Bangla Validation", bangla_validation)
]:

    normalized = (
        df["text"]
        .astype(str)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .str.lower()
    )

    duplicate_group = normalized.duplicated(keep=False)

    if duplicate_group.any():

        review = df.loc[
            duplicate_group, ["text", "label"]
        ].copy()

        review["normalized_text"] = normalized.loc[
            duplicate_group
        ].values

        audit_file = (
            name.lower().replace(" ", "_")
            + "_duplicates.csv"
        )

        review.to_csv(audit_file, index=False)

        # Stop if identical articles have conflicting labels
        conflicts = (
            review.groupby("normalized_text")["label"]
            .nunique()
            .gt(1)
        )

        if conflicts.any():
            raise ValueError(
                f"{name}: conflicting labels detected. "
                f"Review {audit_file} before continuing."
            )

        # Keep the first copy of each article
        remove_mask = normalized.duplicated(keep="first")
        removed = int(remove_mask.sum())

        df.drop(index=df.index[remove_mask], inplace=True)
        df.reset_index(drop=True, inplace=True)

        print(f"{name}: removed {removed} duplicate rows")
        print(f"Audit saved: {audit_file}")

print("\nDataset sizes after standardization:")

for name, df in zip(dataset_names, all_datasets):
    print(f"{name}: {len(df)} rows")

print("\nData types:")
print(english_train.dtypes)
```

The experiment applies a lightweight final text-preparation procedure to the six internal datasets.

Step 20A examines textual noise, including URLs, hashtags, mentions, HTML-like tags, and emojis. The current diagnostic output detects a small number of URLs and hashtags in English validation and additional noise indicators in some Bangla records.

The diagnostic procedure identifies these patterns but does not automatically remove them.

For classical machine learning, separate TF-IDF vectorizers are fitted to English and Bangla training data and subsequently applied to their corresponding validation and test datasets.

The preprocessing and data-distribution differences should be considered methodological limitations until the underlying dataset preparation has been fully standardized and verified.

```python
# Step 20A: Text Noise Diagnostic Check

import string
import re
import pandas as pd

def count_emojis_and_emoticons(text):
    text = str(text)

    # Unicode emoji ranges
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols and pictographs
        "\U0001F680-\U0001F6FF"  # transport and map symbols
        "\U0001F1E0-\U0001F1FF"  # flags
        "\U00002700-\U000027BF"  # dingbats
        "\U00002600-\U000026FF"  # miscellaneous symbols
        "]+",
        flags=re.UNICODE
    )

    # Common text emoticons
    emoticon_pattern = re.compile(
        r"(:-\)|:\)|:-\(|:\(|:D|:-D|;\)|;-\)|:P|:-P|:'\(|😂|🤣|😊|😍|😢|😡)"
    )

    emoji_count = len(emoji_pattern.findall(text))
    emoticon_count = len(emoticon_pattern.findall(text))

    return emoji_count + emoticon_count


def text_noise_profile(df, dataset_name):
    texts = df["text"].astype(str)

    total_rows = len(texts)

    url_rows = texts.str.contains(r"http[s]?://|www\.", regex=True, na=False).sum()
    hashtag_rows = texts.str.contains(r"#\w+", regex=True, na=False).sum()
    mention_rows = texts.str.contains(r"@\w+", regex=True, na=False).sum()
    html_rows = texts.str.contains(r"<[^>]+>", regex=True, na=False).sum()

    emoji_counts = texts.apply(count_emojis_and_emoticons)
    emoji_rows = (emoji_counts > 0).sum()

    punctuation_counts = texts.apply(
        lambda x: sum(1 for ch in str(x) if ch in string.punctuation)
    )

    character_counts = texts.apply(lambda x: max(len(str(x)), 1))
    punctuation_ratio = punctuation_counts / character_counts

    punctuation_heavy_rows = (punctuation_ratio > 0.20).sum()

    return {
        "Dataset": dataset_name,
        "Rows": total_rows,
        "Rows with URLs": int(url_rows),
        "Rows with hashtags": int(hashtag_rows),
        "Rows with @mentions": int(mention_rows),
        "Rows with HTML-like tags": int(html_rows),
        "Rows with emojis/emoticons": int(emoji_rows),
        "Average punctuation ratio": round(float(punctuation_ratio.mean()), 4),
        "Punctuation-heavy rows (>20%)": int(punctuation_heavy_rows)
    }


text_noise_summary = pd.DataFrame([
    text_noise_profile(english_train, "English Train"),
    text_noise_profile(english_validation, "English Validation"),
    text_noise_profile(english_test, "English Test"),
    text_noise_profile(bangla_train, "Bangla Train"),
    text_noise_profile(bangla_validation, "Bangla Validation"),
    text_noise_profile(bangla_test, "Bangla Test")
])

display(text_noise_summary)

text_noise_summary.to_csv("text_noise_summary.csv", index=False)
```

## Text Representation Using TF-IDF

Each news article was treated as a document and converted into a numerical feature vector using TF-IDF. This approach represents text through term-based features and gives higher importance to terms that are frequent in a document but less frequent across the corpus.

Separate TF-IDF vectorizers were used for Bangla and English. This prevented vocabulary mixing between the two languages and allowed each language to maintain its own feature space.

## Hyperparameter Tuning and Cross-Validation

Linear SVM hyperparameter tuning was performed independently for English and Bangla using GridSearchCV on the respective training datasets.

The regularization parameter C was evaluated using the candidate values 0.1, 1, and 10.

A three-fold cross-validation procedure was used for hyperparameter selection, with weighted F1-score as the optimization metric.

The selected C value was 1 for both English and Bangla.

After tuning, a separate five-fold cross-validation analysis was performed on the training datasets using the selected C value to examine model performance across different training folds.

The weighted F1-scores produced by GridSearchCV represent training-set cross-validation performance, not performance on the independent validation datasets.

The independent validation datasets were subsequently used to compare the four classical machine learning classifiers and select the final classical model for each language.

```python

# Step 26: Leakage-Safe Linear SVM Hyperparameter Tuning

from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

svm_params = {"svm__C": [0.1, 1, 10]}

inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

def make_svm_pipeline(lowercase_setting, c=1.0):
    return Pipeline([
        ("tfidf", TfidfVectorizer(max_features=20000, ngram_range=(1, 2), min_df=2,
            max_df=0.95, sublinear_tf=True, lowercase=lowercase_setting)),
        ("svm", LinearSVC(C=c, class_weight="balanced", max_iter=10000, random_state=RANDOM_STATE))
    ])

# English
svm_grid_english = GridSearchCV(estimator=make_svm_pipeline(True),
    param_grid=svm_params, cv=inner_cv,scoring="f1_weighted", n_jobs=-1)

svm_grid_english.fit(X_eng_train, y_eng_train)

print("English best parameters:", svm_grid_english.best_params_)
print("English mean CV weighted F1:", svm_grid_english.best_score_)

# Bangla
svm_grid_bangla = GridSearchCV(estimator=make_svm_pipeline(False),
    param_grid=svm_params, cv=inner_cv,scoring="f1_weighted", n_jobs=-1)

svm_grid_bangla.fit(X_ban_train, y_ban_train)

print("\nBangla best parameters:", svm_grid_bangla.best_params_)
print("Bangla mean CV weighted F1:", svm_grid_bangla.best_score_)
```

```python

# Step 28: Leakage-Safe Five-Fold Cross-Validation

from sklearn.model_selection import cross_val_score, StratifiedKFold

best_c_english = svm_grid_english.best_params_["svm__C"]
best_c_bangla = svm_grid_bangla.best_params_["svm__C"]

outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# English
english_svm_cv = make_svm_pipeline(lowercase_setting=True, c=best_c_english)

english_cv_scores = cross_val_score( english_svm_cv, X_eng_train,
    y_eng_train, cv=outer_cv, scoring="f1_weighted", n_jobs=-1)

print("English 5-Fold CV F1 Scores:", english_cv_scores)
print("English Mean CV F1 Score:", english_cv_scores.mean())

# Bangla
bangla_svm_cv = make_svm_pipeline(lowercase_setting=False, c=best_c_bangla)

bangla_cv_scores = cross_val_score(bangla_svm_cv, X_ban_train, y_ban_train,
    cv=outer_cv, scoring="f1_weighted",n_jobs=-1)

print("\nBangla 5-Fold CV F1 Scores:", bangla_cv_scores)
print("Bangla Mean CV F1 Score:", bangla_cv_scores.mean())
```

## Bias-Variance and Generalization Analysis

To examine whether the trained models suffer from underfitting or overfitting, the training, validation, and test performances were compared. A model may show high variance if it performs very well on the training set but substantially worse on the validation or test sets. A model may show high bias if performance is consistently low across training, validation, and test sets.

In this study, weighted F1-score was used as the main diagnostic metric because the task is a binary classification problem and F1-score balances precision and recall. This analysis helps evaluate whether the models generalize consistently beyond the training data.

```python
# Step 44: Train-Validation-Test Performance Gap Analysis

def compute_split_metrics(model, X_train, y_train, X_val, y_val, X_test, y_test, language, model_name):
    """
    Computes train, validation, and test metrics for generalization analysis.
    This helps identify possible underfitting or overfitting.
    """

    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)
    y_test_pred = model.predict(X_test)

    train_f1 = f1_score(y_train, y_train_pred, average="weighted", zero_division=0)
    val_f1 = f1_score(y_val, y_val_pred, average="weighted", zero_division=0)
    test_f1 = f1_score(y_test, y_test_pred, average="weighted", zero_division=0)

    train_acc = accuracy_score(y_train, y_train_pred)
    val_acc = accuracy_score(y_val, y_val_pred)
    test_acc = accuracy_score(y_test, y_test_pred)

    train_val_gap = train_f1 - val_f1
    train_test_gap = train_f1 - test_f1
    val_test_gap = val_f1 - test_f1

    if train_f1 < 0.70 and val_f1 < 0.70 and test_f1 < 0.70:
        diagnosis = "Possible high bias / underfitting"
    elif train_test_gap > 0.05:
        diagnosis = "Possible high variance / overfitting"
    elif abs(val_test_gap) > 0.05:
        diagnosis = "Possible validation-test distribution difference"
    else:
        diagnosis = "Good generalization"

    return {
        "Language": language,"Model": model_name,
        "Train Accuracy": train_acc,"Validation Accuracy": val_acc, "Test Accuracy": test_acc,
        "Train F1-score": train_f1, "Validation F1-score": val_f1, "Test F1-score": test_f1,
        "Train-Validation F1 Gap": train_val_gap, "Train-Test F1 Gap": train_test_gap,
        "Validation-Test F1 Gap": val_test_gap, "Generalization Diagnosis": diagnosis
    }
bias_variance_results = []

# English models
for model_name, trained_model in english_trained_models.items():
    bias_variance_results.append(
        compute_split_metrics(
            model=trained_model,
            X_train=X_eng_train_tfidf, y_train=y_eng_train,
            X_val=X_eng_val_tfidf, y_val=y_eng_val,
            X_test=X_eng_test_tfidf, y_test=y_eng_test,
            language="English", model_name=model_name
        )
    )

# Bangla models
for model_name, trained_model in bangla_trained_models.items():
    bias_variance_results.append(
        compute_split_metrics(
            model=trained_model,
            X_train=X_ban_train_tfidf, y_train=y_ban_train,
            X_val=X_ban_val_tfidf, y_val=y_ban_val,
            X_test=X_ban_test_tfidf, y_test=y_ban_test,
            language="Bangla", model_name=model_name
        )
    )

bias_variance_summary = pd.DataFrame(bias_variance_results)
bias_variance_summary = bias_variance_summary.round(4)

display(bias_variance_summary)
```

## ROC Curve

ROC-AUC is calculated when model probability scores or decision scores are available. Linear SVM uses a decision function, while Logistic Regression, Naive Bayes, and Random Forest provide probability scores.


```python
# Step 50: ROC Curve Function for Best Models

def get_model_scores(model, X):
    if hasattr(model, "predict_proba"):
        return model.predict_proba(X)[:, 1]
    elif hasattr(model, "decision_function"):
        return model.decision_function(X)
    else:
        return None


def plot_roc_for_model(model, X_test, y_test, language, model_name):
    scores = get_model_scores(model, X_test)
    if scores is None:
        print(f"ROC curve not available for {language} - {model_name}")
        return

    fpr, tpr, _ = roc_curve(y_test, scores)
    auc_value = roc_auc_score(y_test, scores)

    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f"ROC-AUC = {auc_value:.4f}")
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.title(f"ROC Curve: {language} - {model_name}")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()
    plt.show()

plot_roc_for_model(english_trained_models[best_english_model_name], X_eng_test_tfidf, y_eng_test, "English", best_english_model_name)
plot_roc_for_model(bangla_trained_models[best_bangla_model_name], X_ban_test_tfidf, y_ban_test, "Bangla", best_bangla_model_name)
```

## Statistical Significance Testing

McNemar's test was used to compare paired predictions from Linear SVM and Logistic Regression on the same internal test samples.

For English, the comparison produced a p-value of approximately 1.34 × 10^-9, indicating a statistically significant difference in prediction outcomes between Linear SVM and Logistic Regression.

For Bangla, the corresponding p-value was approximately 0.1406, which was not statistically significant at the conventional 0.05 threshold.

These findings indicate that the observed English performance difference between Linear SVM and Logistic Regression was statistically significant in this test set, while the corresponding Bangla difference was not.

Additional pairwise comparisons were conducted against Naive Bayes and Random Forest, with Bonferroni-adjusted p-values reported to account for multiple comparisons.

Statistical significance does not necessarily indicate practical importance or demonstrate generalization beyond the evaluated datasets.

```python
# Step 52: Statistical Significance Test Using McNemar's Test

from statsmodels.stats.contingency_tables import mcnemar

def run_mcnemar_test(y_true, pred_a, pred_b, model_a_name, model_b_name, language):
    y_true_values = np.array(y_true)
    pred_a = np.array(pred_a)
    pred_b = np.array(pred_b)

    correct_a = pred_a == y_true_values
    correct_b = pred_b == y_true_values

    both_correct = ((correct_a == True) & (correct_b == True)).sum()
    a_correct_b_wrong = ((correct_a == True) & (correct_b == False)).sum()
    a_wrong_b_correct = ((correct_a == False) & (correct_b == True)).sum()
    both_wrong = ((correct_a == False) & (correct_b == False)).sum()

    table = [
        [int(both_correct), int(a_correct_b_wrong)],
        [int(a_wrong_b_correct), int(both_wrong)]
    ]

    result = mcnemar(table, exact=False, correction=True)

    print(f"{language}: {model_a_name} vs {model_b_name}")
    print("Contingency Table:", table)
    print("McNemar Statistic:", round(result.statistic, 4))
    print("p-value:", f"{result.pvalue:.4e}")

    if result.pvalue < 0.05:
        print("Interpretation: The difference is statistically significant.\n")
    else:
        print("Interpretation: The difference is not statistically significant.\n")


def get_best_and_second_best_model(final_results_df, language):
    language_results = final_results_df[
        final_results_df["Language"] == language
    ].sort_values(by="F1-score", ascending=False)

    best_model = language_results.iloc[0]["Model"]
    second_best_model = language_results.iloc[1]["Model"]

    return best_model, second_best_model


# English: compare best model with second-best model
english_best_model, english_second_best_model = get_best_and_second_best_model(
    final_results_df,
    "English"
)

run_mcnemar_test(
    y_true=y_eng_test,
    pred_a=english_test_predictions[english_best_model],
    pred_b=english_test_predictions[english_second_best_model],
    model_a_name=english_best_model,
    model_b_name=english_second_best_model,
    language="English"
)


# Bangla: compare best model with second-best model
bangla_best_model, bangla_second_best_model = get_best_and_second_best_model(
    final_results_df,
    "Bangla"
)

run_mcnemar_test(
    y_true=y_ban_test,
    pred_a=bangla_test_predictions[bangla_best_model],
    pred_b=bangla_test_predictions[bangla_second_best_model],
    model_a_name=bangla_best_model,
    model_b_name=bangla_second_best_model,
    language="Bangla"
)
```

## Confidence Interval Analysis

To estimate the uncertainty of the final test results, bootstrap 95% confidence intervals are calculated for the best validation-selected model in each language. This provides a range of plausible values for accuracy and F1-score instead of reporting only a single point estimate.

```python
# Step 53: Bootstrap 95% Confidence Intervals for Best Models

from sklearn.metrics import accuracy_score, f1_score

def bootstrap_metric_ci(y_true, y_pred, metric_function, n_bootstrap=1000, confidence=0.95, random_state=42):
    rng = np.random.default_rng(random_state)
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    scores = []
    n = len(y_true)

    for _ in range(n_bootstrap):
        sample_idx = rng.integers(0, n, n)
        score = metric_function(y_true[sample_idx], y_pred[sample_idx])
        scores.append(score)

    lower = np.percentile(scores, (1 - confidence) / 2 * 100)
    upper = np.percentile(scores, (1 + confidence) / 2 * 100)

    return np.mean(scores), lower, upper

ci_rows = []

for language, y_true, y_pred, model_name in [
    ("English", y_eng_test, english_test_predictions[best_english_validation_model], best_english_validation_model),
    ("Bangla", y_ban_test, bangla_test_predictions[best_bangla_validation_model], best_bangla_validation_model)
]:
    acc_mean, acc_low, acc_high = bootstrap_metric_ci(
        y_true, y_pred, accuracy_score, random_state=RANDOM_STATE
    )

    f1_mean, f1_low, f1_high = bootstrap_metric_ci(
        y_true,
        y_pred,
        lambda yt, yp: f1_score(yt, yp, average="weighted"),
        random_state=RANDOM_STATE
    )

    ci_rows.append({
        "Language": language,
        "Model": model_name,
        "Metric": "Accuracy",
        "Mean": acc_mean,
        "95% CI Lower": acc_low,
        "95% CI Upper": acc_high
    })

    ci_rows.append({
        "Language": language,
        "Model": model_name,
        "Metric": "Weighted F1-score",
        "Mean": f1_mean,
        "95% CI Lower": f1_low,
        "95% CI Upper": f1_high
    })

confidence_interval_table = pd.DataFrame(ci_rows).round(4)
display(confidence_interval_table)
```

## Pairwise Statistical Comparison

To further examine whether the best-performing classifier differs significantly from other classical machine learning models, pairwise McNemar tests were conducted. Linear SVM was compared against Logistic Regression, Naive Bayes, and Random Forest using the same test samples. This analysis helps determine whether observed performance differences are statistically meaningful beyond numerical metric differences.

Because multiple pairwise comparisons are performed, both raw p-values and Bonferroni-adjusted p-values are reported.

```python
# Step 54: Pairwise McNemar Test for Linear SVM Against Other Models

from statsmodels.stats.contingency_tables import mcnemar

def pairwise_mcnemar_table(y_true, predictions_dict, reference_model, language):
    """
    Compare one reference model against all other models using McNemar's test.
    Reports raw p-value and Bonferroni-adjusted p-value.
    """
    results = []

    y_true_values = np.array(y_true)
    reference_pred = np.array(predictions_dict[reference_model])

    comparison_models = [
        model_name for model_name in predictions_dict.keys()
        if model_name != reference_model
    ]

    number_of_comparisons = len(comparison_models)

    for comparison_model in comparison_models:
        comparison_pred = np.array(predictions_dict[comparison_model])

        reference_correct = reference_pred == y_true_values
        comparison_correct = comparison_pred == y_true_values

        both_correct = ((reference_correct == True) & (comparison_correct == True)).sum()
        reference_correct_comparison_wrong = ((reference_correct == True) & (comparison_correct == False)).sum()
        reference_wrong_comparison_correct = ((reference_correct == False) & (comparison_correct == True)).sum()
        both_wrong = ((reference_correct == False) & (comparison_correct == False)).sum()

        contingency_table = [
            [int(both_correct), int(reference_correct_comparison_wrong)],
            [int(reference_wrong_comparison_correct), int(both_wrong)]
        ]

        test_result = mcnemar(
            contingency_table,
            exact=False,
            correction=True
        )

        raw_p_value = test_result.pvalue
        bonferroni_adjusted_p_value = min(raw_p_value * number_of_comparisons, 1.0)

        if bonferroni_adjusted_p_value < 0.05:
            interpretation = "Significant after Bonferroni correction"
        elif raw_p_value < 0.05:
            interpretation = "Significant before correction only"
        else:
            interpretation = "Not statistically significant"

        results.append({
            "Language": language,
            "Reference Model": reference_model,
            "Comparison Model": comparison_model,
            "Both Correct": int(both_correct),
            "Reference Correct / Comparison Wrong": int(reference_correct_comparison_wrong),
            "Reference Wrong / Comparison Correct": int(reference_wrong_comparison_correct),
            "Both Wrong": int(both_wrong),
            "McNemar Statistic": round(test_result.statistic, 4),
            "Raw p-value": round(raw_p_value, 6),
            "Bonferroni-adjusted p-value": round(bonferroni_adjusted_p_value, 6),
            "Interpretation": interpretation
        })

    return pd.DataFrame(results)


# English pairwise comparison: Linear SVM vs other models
english_pairwise_mcnemar = pairwise_mcnemar_table(
    y_true=y_eng_test,
    predictions_dict=english_test_predictions,
    reference_model="Linear SVM",
    language="English"
)

# Bangla pairwise comparison: Linear SVM vs other models
bangla_pairwise_mcnemar = pairwise_mcnemar_table(
    y_true=y_ban_test,
    predictions_dict=bangla_test_predictions,
    reference_model="Linear SVM",
    language="Bangla"
)

# Combine English and Bangla results
pairwise_mcnemar_results = pd.concat(
    [english_pairwise_mcnemar, bangla_pairwise_mcnemar],
    ignore_index=True
)

display(pairwise_mcnemar_results)
```

## Error Category Analysis

A rule-based error categorization procedure was applied to explore observable characteristics of the articles misclassified by the validation-selected classical models.

The analysis considers categories such as:

- Sensational or viral-style wording
- Political or named-entity-heavy content
- Very long articles
- Short text or limited context
- General lexical or semantic ambiguity

These categories were assigned automatically using predefined heuristic rules.

They do not represent independently verified linguistic annotations or confirmed explanations for individual misclassifications.

The resulting distributions are therefore descriptive and exploratory.

Manual qualitative analysis and validated error-annotation procedures would be needed to establish the underlying reasons for prediction errors.

```python
# Step 56: Rule-Based Error Category Analysis

def assign_error_category(text):
    text = str(text).lower()
    word_count = len(text.split())

    political_terms = [
        "trump", "biden", "election", "government", "minister", "president",
        "নির্বাচন", "সরকার", "মন্ত্রী", "প্রধানমন্ত্রী", "আওয়ামী", "বিএনপি"
    ]

    sensational_terms = [
        "shocking", "breaking", "viral", "exposed", "secret", "watch",
        "ভাইরাল", "অবিশ্বাস্য", "চাঞ্চল্যকর", "গোপন", "দেখুন"
    ]

    if word_count < 50:
        return "Short text / limited context"
    elif any(term in text for term in sensational_terms):
        return "Sensational or viral-style wording"
    elif any(term in text for term in political_terms):
        return "Political or named-entity-heavy content"
    elif word_count > 500:
        return "Very long article"
    else:
        return "General lexical/semantic ambiguity"

def summarize_error_categories(error_df, language):
    temp = error_df.copy()
    temp["Error Category"] = temp["Text"].apply(assign_error_category)

    summary = (
        temp.groupby("Error Category")
        .size()
        .reset_index(name="Number of Errors")
        .sort_values("Number of Errors", ascending=False)
    )

    summary["Language"] = language
    return summary[["Language", "Error Category", "Number of Errors"]]

english_error_category_summary = summarize_error_categories(english_error_df, "English")
bangla_error_category_summary = summarize_error_categories(bangla_error_df, "Bangla")

error_category_summary = pd.concat(
    [english_error_category_summary, bangla_error_category_summary],
    ignore_index=True
)

display(error_category_summary)
```

```python
# Step 57: Error Type Summary

def error_type_summary(error_df, language):
    if error_df.empty:
        return pd.DataFrame({"Language": [language], "Error Type": ["No errors"], "Count": [0]})

    error_df = error_df.copy()
    error_df["Error Type"] = error_df.apply(
        lambda row: "Fake predicted as Real" if row["Actual Label"] == 0 and row["Predicted Label"] == 1 else "Real predicted as Fake",
        axis=1
    )

    summary = error_df["Error Type"].value_counts().reset_index()
    summary.columns = ["Error Type", "Count"]
    summary.insert(0, "Language", language)
    return summary

error_summary_table = pd.concat([
    error_type_summary(english_error_df, "English"),
    error_type_summary(bangla_error_df, "Bangla")
], ignore_index=True)

display(error_summary_table)
```

## TF-IDF Feature Interpretation

For linear models such as Logistic Regression and Linear SVM, feature weights can be inspected to identify n-grams associated with each class. These features should be interpreted carefully as model-associated indicators, not absolute evidence of truth or falsehood.


```python
# Step 58: Top TF-IDF Features for Logistic Regression

def show_top_logistic_features(model, vectorizer, language, top_n=20):
    if not hasattr(model, "coef_"):
        print(f"{language}: This model does not provide linear coefficients.")
        return None

    feature_names = np.array(vectorizer.get_feature_names_out())
    coefs = model.coef_[0]

    top_real_idx = np.argsort(coefs)[-top_n:][::-1]
    top_fake_idx = np.argsort(coefs)[:top_n]

    feature_df = pd.DataFrame({
        "Language": [language] * (top_n * 2),
        "Class Association": ["Real News (1)"] * top_n + ["Fake News (0)"] * top_n,
        "Feature": list(feature_names[top_real_idx]) + list(feature_names[top_fake_idx]),
        "Coefficient": list(coefs[top_real_idx]) + list(coefs[top_fake_idx])
    })

    display(feature_df)
    return feature_df

english_lr_model = english_trained_models["Logistic Regression"]
bangla_lr_model = bangla_trained_models["Logistic Regression"]

english_top_features = show_top_logistic_features(english_lr_model, english_tfidf, "English", top_n=20)
bangla_top_features = show_top_logistic_features(bangla_lr_model, bangla_tfidf, "Bangla", top_n=20)
```

## TF-IDF Ablation Study

An ablation study was conducted to examine how different TF-IDF configurations affect fake news detection performance. The purpose of this experiment is to evaluate whether unigram features alone are sufficient or whether adding bigram features improves model performance.

To keep the ablation focused and computationally efficient, Linear SVM was used as the classifier because it showed strong performance in the main experiments. The ablation was performed on the validation set only, while the test set remained reserved for final model reporting.

```python
# Step 60: TF-IDF Ablation Study Using Linear SVM

def run_tfidf_ablation(
    X_train,
    y_train,
    X_val,
    y_val,
    language,
    best_c,
    lowercase_setting
):
    """
    Runs TF-IDF ablation experiments using Linear SVM.
    The validation set is used for comparison.
    The test set is not used in this ablation to avoid test-set overuse.
    """

    ablation_configs = [{
            "Configuration": "Unigram, 10k features",
            "ngram_range": (1, 1),
            "max_features": 10000
        },
        {
            "Configuration": "Unigram, 20k features",
            "ngram_range": (1, 1),
            "max_features": 20000
        },
        {
            "Configuration": "Unigram + Bigram, 10k features",
            "ngram_range": (1, 2),
            "max_features": 10000
        },
        {
            "Configuration": "Unigram + Bigram, 20k features",
            "ngram_range": (1, 2),
            "max_features": 20000
        }
    ]

    ablation_results = []
    for config in ablation_configs:
        vectorizer = TfidfVectorizer(
            max_features=config["max_features"],
            ngram_range=config["ngram_range"],
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            lowercase=lowercase_setting
        )

        X_train_tfidf_ablation = vectorizer.fit_transform(X_train)
        X_val_tfidf_ablation = vectorizer.transform(X_val)
        model = LinearSVC(
            C=best_c,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            max_iter=10000
        )

        model.fit(X_train_tfidf_ablation, y_train)
        y_val_pred = model.predict(X_val_tfidf_ablation)
        ablation_results.append({
            "Language": language,
            "Configuration": config["Configuration"],
            "ngram_range": str(config["ngram_range"]),
            "max_features": config["max_features"],
            "Validation Accuracy": accuracy_score(y_val, y_val_pred),
            "Validation Precision": precision_score(y_val, y_val_pred, average="weighted", zero_division=0),
            "Validation Recall": recall_score(y_val, y_val_pred, average="weighted", zero_division=0),
            "Validation F1-score": f1_score(y_val, y_val_pred, average="weighted", zero_division=0),
            "Validation Macro-F1": f1_score(y_val, y_val_pred, average="macro", zero_division=0)
        })
    return pd.DataFrame(ablation_results)

# English TF-IDF ablation
english_ablation_df = run_tfidf_ablation(
    X_train=X_eng_train,
    y_train=y_eng_train,
    X_val=X_eng_val,
    y_val=y_eng_val,
    language="English",
    best_c=best_c_english,
    lowercase_setting=True
)
# Bangla TF-IDF ablation
bangla_ablation_df = run_tfidf_ablation(
    X_train=X_ban_train,
    y_train=y_ban_train,
    X_val=X_ban_val,
    y_val=y_ban_val,
    language="Bangla",
    best_c=best_c_bangla,
    lowercase_setting=False
)
# Combine ablation results
ablation_results_df = pd.concat(
    [english_ablation_df, bangla_ablation_df],
    ignore_index=True
)

ablation_results_df = ablation_results_df.round(4)
display(ablation_results_df)
```

```python
# Step 65: Prepare Dataset for Transformer Models

def prepare_transformer_dataframe(df):
    df = df.copy()
    df = df[["text", "label"]].dropna()
    df["text"] = df["text"].astype(str).str.strip()
    df = df[df["text"] != ""]
    df["label"] = df["label"].astype(int)
    df = df[df["label"].isin([0, 1])]
    return df.reset_index(drop=True)

bangla_train_tf = prepare_transformer_dataframe(bangla_train)
bangla_validation_tf = prepare_transformer_dataframe(bangla_validation)
bangla_test_tf = prepare_transformer_dataframe(bangla_test)

english_train_tf = prepare_transformer_dataframe(english_train)
english_validation_tf = prepare_transformer_dataframe(english_validation)
english_test_tf = prepare_transformer_dataframe(english_test)

print("Bangla train:", bangla_train_tf.shape)
print("Bangla validation:", bangla_validation_tf.shape)
print("Bangla test:", bangla_test_tf.shape)

print("English train:", english_train_tf.shape)
print("English validation:", english_validation_tf.shape)
print("English test:", english_test_tf.shape)
```

## Transformer Training Function

A reusable training function is used for all transformer models. The function performs tokenization, model fine-tuning, validation, final test evaluation, and metric reporting.

```python
# Step 66: Transformer Training and Evaluation Function

def compute_transformer_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=1)

    return {
        "accuracy": accuracy_score(labels, predictions),
        "precision_weighted": precision_score(labels, predictions, average="weighted", zero_division=0),
        "recall_weighted": recall_score(labels, predictions, average="weighted", zero_division=0),
        "f1_weighted": f1_score(labels, predictions, average="weighted", zero_division=0),
        "f1_macro": f1_score(labels, predictions, average="macro", zero_division=0)
    }


def run_transformer_experiment(
    model_name,
    train_df,validation_df,test_df,
    language,
    experiment_name,
    max_length=256,
    num_train_epochs=2,
    batch_size=8,
    learning_rate=2e-5
):
    print("=" * 80)
    print(f"Running experiment: {experiment_name}")
    print(f"Language: {language}")
    print(f"Model: {model_name}")
    print("=" * 80)

    tokenizer = AutoTokenizer.from_pretrained(model_name)

    train_dataset = Dataset.from_pandas(train_df)
    validation_dataset = Dataset.from_pandas(validation_df)
    test_dataset = Dataset.from_pandas(test_df)

    def tokenize_function(batch):
        return tokenizer(
            batch["text"],
            truncation=True,
            max_length=max_length
        )

    train_dataset = train_dataset.map(tokenize_function, batched=True)
    validation_dataset = validation_dataset.map(tokenize_function, batched=True)
    test_dataset = test_dataset.map(tokenize_function, batched=True)

    train_dataset = train_dataset.remove_columns(["text"])
    validation_dataset = validation_dataset.remove_columns(["text"])
    test_dataset = test_dataset.remove_columns(["text"])

    train_dataset.set_format("torch")
    validation_dataset.set_format("torch")
    test_dataset.set_format("torch")

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name,
        num_labels=2,
        ignore_mismatched_sizes=True
    )

    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

    output_dir = f"./transformer_output_{experiment_name.replace(' ', '_').replace('/', '_')}"

    training_args = TrainingArguments(
        output_dir=output_dir,
        eval_strategy="epoch",
        save_strategy="epoch",
        learning_rate=learning_rate,
        per_device_train_batch_size=batch_size,
        per_device_eval_batch_size=batch_size,
        num_train_epochs=num_train_epochs,
        weight_decay=0.01,
        load_best_model_at_end=True,
        metric_for_best_model="f1_weighted",
        greater_is_better=True,
        logging_steps=100,
        save_total_limit=1,
        report_to="none",
        seed=RANDOM_STATE
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=validation_dataset,
        data_collator=data_collator,
        compute_metrics=compute_transformer_metrics
    )

    trainer.train()

    test_output = trainer.predict(test_dataset)
    test_logits = test_output.predictions
    test_predictions = np.argmax(test_logits, axis=1)
    y_true = test_df["label"].values

    result = {
        "Language": language,
        "Experiment": experiment_name,
        "Model": model_name,
        "Test Accuracy": accuracy_score(y_true, test_predictions),
        "Test Precision": precision_score(y_true, test_predictions, average="weighted", zero_division=0),
        "Test Recall": recall_score(y_true, test_predictions, average="weighted", zero_division=0),
        "Test F1-score": f1_score(y_true, test_predictions, average="weighted", zero_division=0),
        "Test Macro-F1": f1_score(y_true, test_predictions, average="macro", zero_division=0)
    }

    report = classification_report(
        y_true,
        test_predictions,
        target_names=["Fake News (0)", "Real News (1)"],
        zero_division=0
    )

    cm = confusion_matrix(y_true, test_predictions)

    print("\nTest Result:")
    print(result)

    print("\nClassification Report:")
    print(report)

    print("\nConfusion Matrix:")
    print(cm)

    del model
    del trainer
    torch.cuda.empty_cache()
    gc.collect()

    return result, test_predictions
```

## Limitations

This study has several limitations that should be considered when interpreting the experimental findings.

First, although the internal dataset contains 40,064 records after processing, the modeling files do not consistently preserve complete record-level source information, including original publication URLs, source-row identifiers, and labeling histories.

Second, the English validation dataset contains 2,024 records after removing 16 duplicate articles. Although the dataset is substantially larger than in earlier experimental versions, its composition and preprocessing characteristics may differ from the English training and testing datasets.

Third, the English training and testing text appears substantially more stemmed or normalized than the English validation text. This discrepancy may influence feature matching, model selection, and comparisons between validation and test performance.

Fourth, the notebook detected no exact or normalized text overlap between internal splits. However, these checks cannot completely exclude semantically similar articles, reports of the same event, or source-level dependencies.

Fifth, the models demonstrated differences between validation and test performance, particularly in English. These differences complicate interpretation of model selection and generalization.

Sixth, external evaluation of the validation-selected Logistic Regression models showed lower weighted F1-scores than internal testing. English performance decreased from 0.9485 to 0.8355, while Bangla performance decreased from 0.8025 to 0.6179.

Seventh, external evaluation was not performed for every classical and transformer model in the recorded experiments. Consequently, the relative external robustness of all evaluated approaches remains undetermined.

Eighth, transformer experiments were conducted with two training epochs and a maximum input length of 256 tokens. These settings may limit the use of information from longer articles, and more extensive hyperparameter optimization could produce different results.

Ninth, the reported transformer results do not establish training stability across multiple random seeds.

Finally, the classification models learn patterns associated with the provided dataset labels. They do not independently verify factual claims through external evidence, source authentication, or automated fact-checking.

The findings should therefore be interpreted as results of supervised news-text classification under the evaluated dataset conditions rather than as proof of comprehensive factual verification capability.

## Methodology Summary for Thesis Writing

This study investigated automated fake news detection in Bangla and English using classical machine learning and transformer-based NLP approaches.

### Dataset Preparation

Separate English and Bangla datasets were used for training, validation, and internal testing.

The six original internal CSV files contained 40,080 records. After removing 16 duplicate English validation records, the final modeling dataset contained 40,064 records, including 20,024 English and 20,040 Bangla records.

For each language, 14,000 records were used for training and 4,000 for internal testing.

The final English validation dataset contained 2,024 records, while Bangla validation contained 2,040 records.

The label mapping was consistent throughout the experiment:

- 0 = Fake News
- 1 = Real News

Training and testing datasets were class-balanced for both languages.

### Data Quality Control

The implemented checks included missing-value detection, empty-text identification, binary-label verification, within-split duplicate checks, and exact and normalized cross-split overlap analysis.

No remaining exact or normalized duplicate texts or cross-split overlaps were detected by the implemented procedures.

However, the checks do not establish the absence of semantic duplicates or source-level dependencies.

### Classical Machine Learning

Separate TF-IDF vectorizers were used for English and Bangla.

Each vectorizer was fitted using the corresponding training data and then applied to validation and testing data.

Four classifiers were evaluated:

1. Logistic Regression
2. Multinomial Naive Bayes
3. Linear SVM
4. Random Forest

Linear SVM was tuned using three-fold GridSearchCV, followed by a separate five-fold training-set cross-validation analysis.

Classical model selection was performed using independent validation weighted F1-score.

Logistic Regression achieved the highest classical validation weighted F1-score for both languages:

- English validation F1: 0.7190
- Bangla validation F1: 0.8284

The corresponding internal test weighted F1-scores were 0.9485 and 0.8025.

Among all classical classifiers, Linear SVM achieved the highest observed internal test weighted F1-scores:

- English: 0.9645
- Bangla: 0.8097

### Transformer Experiments

Three transformer architectures were evaluated:

- BanglaBERT for Bangla
- Multilingual BERT for English and Bangla
- XLM-RoBERTa for English and Bangla

The recorded fine-tuning configuration used two training epochs and a maximum input sequence length of 256 tokens.

For English, mBERT achieved a weighted F1-score of 0.9510, while XLM-RoBERTa achieved 0.9278.

For Bangla, BanglaBERT achieved 0.8460, mBERT achieved 0.8101, and XLM-RoBERTa achieved 0.7866.

BanglaBERT achieved the highest observed Bangla internal test weighted F1-score among all evaluated models.

Linear SVM achieved the highest observed English internal test weighted F1-score.

### Evaluation and Statistical Analysis

Model performance was evaluated using accuracy, precision, recall, weighted F1-score, macro-F1, and confusion matrices.

ROC-AUC was calculated for the classical classifiers using appropriate probability or decision scores.

Additional analyses included McNemar statistical comparisons, bootstrap confidence intervals, TF-IDF feature interpretation, rule-based error categorization, and TF-IDF feature-configuration ablation experiments.

### External Evaluation

The validation-selected Logistic Regression models were evaluated on separate external datasets:

- English: 68,002 records
- Bangla: 18,150 records

English Logistic Regression achieved an external accuracy of 0.8370 and weighted F1-score of 0.8355.

Bangla Logistic Regression achieved an external accuracy of 0.6212 and weighted F1-score of 0.6179.

Both external weighted F1-scores were lower than their corresponding internal test results.

### Overall Methodological Interpretation

This study provides a comparative evaluation of classical and transformer-based fake news detection approaches across English and Bangla.

The recorded results suggest that Linear SVM is competitive for English internal classification, while BanglaBERT achieved the highest Bangla internal test performance.

However, differences in preprocessing, dataset distributions, source provenance, and external evaluation performance limit stronger conclusions about model generalization.

Further data verification and robustness experiments would strengthen the reliability of the findings.

```python
# Step 77: Validation Dataset Duplicate Row Inspection

import pandas as pd
from IPython.display import display

print("=" * 65)
print("STEP 77: VALIDATION DATASET DUPLICATE ROW INSPECTION")
print("=" * 65)

validation_files = {"English": found_paths["english_validation"],"Bangla": found_paths["bangla_validation"]}

for language, file_path in validation_files.items():
    print(f"\n{'=' * 65}")
    print(f"{language.upper()} VALIDATION DATASET")
    print("=" * 65)

    # Reload original CSV to preserve original row numbers
    df = pd.read_csv(file_path)

    # Original row positions
    df["Original Row Number"] = df.index + 1
    df["CSV Row Number"] = df.index + 2

    # Apply the same validity checks as Step 9
    labels = pd.to_numeric(df["label"], errors="coerce")

    valid_text = (
        df["text"]
        .astype("string")
        .str.strip()
        .ne("")
        .fillna(False)
    )

    valid_rows = labels.isin([0, 1]) & valid_text
    invalid_count = int((~valid_rows).sum())
    df = df.loc[valid_rows].copy()

    # Same text normalization used in Step 9
    df["normalized_text"] = (
        df["text"]
        .astype(str)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .str.lower()
    )

    # Identify all occurrences of duplicated articles
    duplicate_mask = df["normalized_text"].duplicated(
        keep=False
    )

    duplicates = df.loc[duplicate_mask].copy()

    if duplicates.empty:
        print("No duplicate articles found.")
        continue

    # Assign a group ID to each duplicate article
    duplicates["Duplicate Group"] = (
        pd.factorize(duplicates["normalized_text"])[0] + 1
    )

    # Show kept and removed records
    duplicates["Status"] = (
        duplicates["normalized_text"]
        .duplicated(keep="first")
        .map({
            False: "Kept - First Occurrence",
            True: "Removed - Duplicate"
        })
    )

    # Extract first non-empty line
    def get_first_line(text):
        lines = str(text).splitlines()

        for line in lines:
            if line.strip():
                return line.strip()

        return ""

    duplicates["First Line"] = (
        duplicates["text"].apply(get_first_line)
    )

    # Short preview for notebook display
    duplicates["Text Preview"] = (duplicates["First Line"].str.slice(0, 200))

    # Sort duplicate groups
    duplicates = duplicates.sort_values(
        ["Duplicate Group", "Original Row Number"]
    )

    # Summary
    removed_count = (
        duplicates["Status"] == "Removed - Duplicate"
    ).sum()

    print(f"Invalid rows excluded from check: {invalid_count}")
    print(f"Duplicate groups: {duplicates['Duplicate Group'].nunique()}")
    print(f"Total rows in duplicate groups: {len(duplicates)}")
    print(f"Duplicate rows removed in Step 9: {removed_count}")

    # Display results
    display_columns = [
        "Duplicate Group",
        "Original Row Number",
        "CSV Row Number",
        "label",
        "Status",
        "Text Preview"
    ]

    print("\nDuplicate Article Details:")

    display(
        duplicates[display_columns]
        .reset_index(drop=True)
    )

    # Save report containing the complete first line
    save_columns = ["Duplicate Group", "Original Row Number", "CSV Row Number", "label", "Status", "First Line", "text"]

    output_file = (
        f"{language.lower()}_validation_duplicate_details.csv"
    )

    duplicates[save_columns].to_csv(output_file, index=False, encoding="utf-8-sig")

    print(f"\nReport saved: {output_file}")

print("\nStep 77 completed successfully.")
```
